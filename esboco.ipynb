```python
file = "Gracie Abrams - Good Reason (Official Lyric Video) - Gracie Abrams (128k).mp3"
```

```python
import whisper 

lyric_model = whisper.load_model("turbo", device = "cuda:0")
```

```python
import subprocess

# Executa o comando 'ls -l' (Linux/Mac) ou 'dir' (Windows)
# O argumento 'shell=True' permite passar o comando como texto único
r = subprocess.run(f'demucs "{file}"', shell = True)

```

```python
lyric_result = lyric_model.transcribe(f"separated/htdemucs/{file.rsplit('.')[0]}/vocals.wav", word_timestamps = True)
```

```python
lyric_result
```

```python
lyrics = []
for i, segment in enumerate(lyric_result["segments"]):
    for word in segment["words"]:
        single_word = word["word"].strip()
        # timestamp = word["start"]
        lyrics.append({
            "single_word": single_word, 
            # "timestamp": timestamp, 
            "start": word["start"],
            "end": word["end"],
            "verse": i
        })
```

```python
lyrics
```

```python
from transformers import AutoModel
chord_model = AutoModel.from_pretrained("m-a-p/SheetSage2", trust_remote_code = True, device_map = "auto")
```

```python
chord_result = chord_model.transcribe(file, output_dir = "output")
```

```python
chords = []
for event in chord_result["events"]:
    if "chord" in event["values"].keys() and event["values"]["chord"] == "N":
        continue
    if "chord" in event["values"].keys() and "timestamp" in event["values"].keys():
        chord = event["values"]["chord"]
        timestamp = event["values"]["timestamp"]

        chord = chord.replace(":maj", "").replace(":min", "m")

        chords.append({
            "chord": chord, 
            "timestamp": timestamp
        })
```

```python
from collections import defaultdict

verses = defaultdict(list)

for word in lyrics:
    verses[word["verse"]].append(word)
```

```python
# Introdução

first_lyric_time = min(
    word["start"]
    for word in lyrics
)

intro_chords = [
    chord for chord in chords
    if chord["timestamp"] < first_lyric_time
]

filtered_intro_chords = []

for chord in intro_chords:

    if (
        not filtered_intro_chords
        or chord["chord"] != filtered_intro_chords[-1]["chord"]
    ):
        filtered_intro_chords.append(chord)

if filtered_intro_chords:

    print("[Intro]")

    print(
        " ".join(
            chord["chord"]
            for chord in filtered_intro_chords
        )
    )

    print()
```

```python
# Une letra com os acordes
for verse_number, words in verses.items():

    if not words:
        continue

    start_time = words[0]["start"]

    if verse_number + 1 in verses:
        end_time = verses[verse_number + 1][0]["start"]
    else:
        end_time = words[-1]["end"]

    verse_chords = [
        chord for chord in chords
        if start_time <= chord["timestamp"] < end_time
    ]

    # Monta a linha da letra

    lyric_line = ""
    word_positions = []

    for word in words:

        if lyric_line:
            lyric_line += " "

        position = len(lyric_line)

        word_positions.append({
            "word": word["single_word"],
            "position": position,
            "start": word["start"],
            "end": word["end"]
        })

        lyric_line += word["single_word"]

    # Descobre a posição de cada acorde

    chord_positions = []

    for chord in verse_chords:

        timestamp = chord["timestamp"]

        target_word = None

        for word in word_positions:

            if word["start"] <= timestamp <= word["end"]:
                target_word = word
                break

        if target_word is None:

            target_word = min(
                word_positions,
                key = lambda word: min(
                    abs(timestamp - word["start"]),
                    abs(timestamp - word["end"])
                )
            )

        chord_positions.append({
            "chord": chord["chord"],
            "timestamp": timestamp,
            "word": target_word
        })


    # Calcula a posição horizontal de cada acorde

    final_chords = []

    for item in chord_positions:

        chord = item["chord"]
        timestamp = item["timestamp"]
        word = item["word"]

        word_start = word["start"]
        word_end = word["end"]

        text_start = word["position"]

        word_length = len(word["word"])

        duration = word_end - word_start

        if duration > 0:

            relative_position = (
                (timestamp - word_start) / duration
            )

            relative_position = max(
                0,
                min(1, relative_position)
            )

        else:
            relative_position = 0

        offset = round(
            relative_position * word_length
        )

        position = text_start + offset

        final_chords.append({
            "chord": chord,
            "position": position,
            "timestamp": timestamp
        })

    final_chords.sort(
        key = lambda x: x["timestamp"]
    )


    # Evita que acordes sejam sobrescritos

    last_end = -1

    for item in final_chords:

        position = item["position"]
        chord = item["chord"]

        if position <= last_end:

            position = last_end + 1

            item["position"] = position

        last_end = position + len(chord) - 1


    # Monta linha dos acordes

    if final_chords:

        required_size = max(
            item["position"] + len(item["chord"])
            for item in final_chords
        )

    else:
        required_size = 0


    line_size = max(
        len(lyric_line),
        required_size
    ) + 5

    chord_line = [" "] * line_size


    for item in final_chords:

        chord = item["chord"]
        position = item["position"]

        for i, char in enumerate(chord):

            chord_line[position + i] = char


    print("".join(chord_line).rstrip())
    print(lyric_line)
```
